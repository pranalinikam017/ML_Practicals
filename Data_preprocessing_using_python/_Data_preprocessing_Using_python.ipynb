```python
#Outlier detection in Age column = IQR
```

```python
# Convert to a standard Python integer
q1 = float(df['Age'].quantile(0.25))
print(q1)  # Output: 22

```

```python
q3 = float(df['Age'].quantile(0.75))
print(q3)
```

```python
iqr=q3-q1
iqr
```

```python
lower=q1-1.5*iqr
lower
```

```python
upper=q3+1.5*iqr
upper
```

```python
df[(df['Age']<lower) & (df['Age']>upper)]
```

```python
q1=float(df['Salary'].quantile(0.25))
q1
```

```python
q3=float(df['Salary'].quantile(0.75))
q3
```

```python
iqr=q3-q1
iqr
```

```python
lower=q1-1.5*iqr
lower
```

```python
upper=q3+1.5*iqr
upper
```

```python
df[(df['Salary']<lower) & (df['Salary']>upper)]
#no outliers detected using IQR.
```
